```python
#created a helper function for finding cost of a path using bfs
def bfs_cost(starting_cell, goal_cell, legal_moves):

    queue = [(starting_cell, 0)] #mod 1: keep track of cost with a tuple (cell, cost)
    visited = []
 
    while queue:
 
        current_cell, cost = queue.pop(0)

        if current_cell not in visited:
            visited.append(current_cell)

            if current_cell == goal_cell:
                return cost #return cost rather than visited cells

            for neighbour in legal_moves[current_cell]:
                if neighbour not in visited:
                    queue.append((neighbour, cost + 1))  #we are appending the neighbour with an incremented cost
def path_cost(start_cell, goal_cell, legal_moves): #lets use bfs from assignemnt 1 since that is the shorter path 
    # Finish the function
    cost= {} #created a dictionary of the total costs of each cell from start to goal
    cost_start_cell = bfs_cost(start_cell, goal_cell, legal_moves) #get the cost from start to goal
    cost[start_cell] = cost_start_cell #store the cost in the dictionary with goal
    for neighbor in legal_moves[start_cell]: #loop through the neighbors of the goal cell
        cost_neighbor = bfs_cost(neighbor, goal_cell, legal_moves) #get the cost from start to neighbor
        cost[neighbor] = cost_neighbor #store the cost in the dictionary with neighbor
    return cost #return path_cost of start cell to goal cell and its neighbors
```

```python
#notes: for each legal next step, we need estimated distance from that cell to the goal cell
#when they say we want to think about the relaxed version of the game, they mean no obstacles) to solve this problem
#manhattan and euclidean distances: complete the functions below according to their instructions
#third task: there was a notebook we used in class for euclidean distance that must modified
    #Don't write a new function for this task

import math # importing math module to use the sqrt function for calculating Euclidean distance

def euclideanDistance(current_cell, all_cells, goal_cell, cellSize, obstacles):
    ''' Compute the Euclidean distance between every legal next cell and the goal cell.
    Since this is meant to act as a heuristic (estimation of the actual distance), obstacles are not
    taken into consideration as "obstacles".
    current_cell : the coordinates of the current location of the robot
    all_cells : the coordinates of all the cells in the grid
    goal_cell : the coordinates of the target cell
    cellSize : the size of a cell in the grid
    obstacles : the coordinates of the cells that are obstacles
    
    The function should return a dictionary where key,value is cell, distance of cell to goal cell, sorted in ascending order.
    '''
    x, y = current_cell
    xG, yG = goal_cell
    eDistances = {}
    # Finish the function
    # Compute the distance between the current cell and the goal cell
    # Then, compute the distance between every legal neighbour of the current cell and the goal cell
    # Store these in the eDistances dictionary and return it sorted in ascending order of the distances
    
    #Kylie
     #current cell to goal via distance formula
    eDistances[current_cell] = math.sqrt((x - xG)**2 + (y - yG)**2) 

    #find distance from every legal neighbour to goal

    #loop through the four possible moves (up, down, left, right)
    for dx,dy in [(0, -cellSize),(0,cellSize),(-cellSize, 0),(cellSize, 0)]:
        #calculate the coordinates of the neighbour cell
        n= (x +dx, y +dy)
        if n in all_cells and n not in obstacles: 
            #calculate the distance from the neighbour cell to the goal cell using the distance formula
            eDistances[n] = math.sqrt((n[0]- xG)**2 + (n[1] -yG)**2)   # obstacles are ignored 

    #returning a sorted dictionary of distances for the current cell and its legal neighbours to the goal cell
    return dict(sorted(eDistances.items(),key=lambda item: item[1]))
    #key lambda item is sorting the dictionary in ascending order based on the distance values (item[1]) of the dictionary items

def manhattanDistance(current_cell,all_cells,goal_cell, cellSize,obstacles):

    ''' Exactly the same as in the euclideanDistance function. 
       The only difference is that now you need to compute the Manhattan Distance. 
       The rest of the code would be pretty much the same as in the previous function.
       '''
   #so we copied the euclidanDistance function and 
   #modified it to compute the Manhattan distance instead of the Euclidean distance
    x,y =current_cell
    xG,yG =goal_cell
    mDistances = {}

    mDistances[current_cell]=abs(x-xG) + abs(y - yG) #Mod: Manhattan distance from current cell to goal
    for dx,dy in [(0,-cellSize), (0,cellSize), (-cellSize, 0),(cellSize, 0)]:
        n = (x +dx, y + dy)
        if n in all_cells and n not in obstacles:
            mDistances[n] = abs(n[0]- xG) + abs(n[1]- yG)  #Mod: Manhattan distance from current cell to goal

    #again we are returning the sorted dictionary of distances for the current cell and its legal neighbours to the goal cell in ascending order.
    return dict(sorted(mDistances.items(),key=lambda item: item[1]))



#Write code for checking admissibility for a given heuristic for the game
def is_admissible(heuristic, all_cells, obstacles, goal_cell, cellSize, legal_moves):
    #parameters: heuristic function, all cells in the grid, obstacles, goal cell, cell size, legal moves
    
    #note: instead of looping through a graph, 
    # we are looping through all the cells in the grid and checking if the 
    # heuristic value is less than or equal to the actual cost of reaching the goal cell from that cell. 
    # If it is not, then the heuristic is not admissible.

    #we also passed in heuristic function as a parameter and use it to find value for the current cell 
    #this can be either the euclidean distance or manhattan distance function. 
    #looping through all the cells in the grid 
    for cell in all_cells: 
        #if the cell is an obstacle, we skip it and continue to the next cell
        if cell in obstacles:
            continue

        #calculate the heuristic value for the current cell using the provided heuristic function
        heuristicEstimate = heuristic(cell, all_cells, goal_cell, cellSize, obstacles)[cell]
        actual_cost = path_cost(cell, goal_cell, legal_moves)[cell] # calculate the actual cost of the shortest path from the current cell to the goal cell using the path_cost function

        #Ureachable cells have no finite actual path cost to compare.
        if actual_cost is not None and heuristicEstimate > actual_cost:
            #the heuristic is not admissable when the estimate > actual cost
            print("Not admissible at", cell, ": heuristic estimate: =", heuristicEstimate, ">= actual =", actual_cost)
            return False
    return True




# Write code for checking consistency for a given heuristic for the game
def is_consistent(heuristic, all_cells, obstacles, goal_cell, cellSize, legal_moves):
    #parameters: heuristic function, all cells in the grid, obstacles, goal cell, cell size, legal moves
    #looping through all the cells in the grid
    for cell in all_cells:
        if cell in obstacles:
            continue
        #calculate the heuristic value for the current cell using the provided heuristic function
        hCell = heuristic(cell, all_cells, goal_cell, cellSize, obstacles)[cell]
        
        #loop through all the legal moves from the current cell
        for neighbor in legal_moves[cell]:
            hNeighbor = heuristic(
                neighbor, all_cells, goal_cell, cellSize, obstacles
            )[neighbor]

            #Each legal move has a cost of 1.
            #if the heuristic value of the current cell is greater than the heuristic value of the neighbor cell plus 1, then the heuristic is not consistent
            if hCell > 1 + hNeighbor:
                print("Not consistent for", cell, "->", neighbor)
                return False
            #note consistency means that the heuristic value of a cell should 
            #not be greater than the heuristic value of any of its neighbors plus
            #the cost of moving to that neighbor (which is 1 in this case).
    return True
```
